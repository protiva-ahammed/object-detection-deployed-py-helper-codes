```python
import cv2
import os

image_path = "./archive/images/train/IMG_0710.jpg"
output_dir = "./archive/tfresized_data/images/train"

def resize_image(image_path,outdupt_dir,target_size=(300,300)):
    image = cv2.imread(image_path)
    if image is None:
        print(f"Failed to load images  for path {image_path}")
        return
    resized_image = cv2.resize(image,target_size)
    image_name = os.path.basename(image_path)
    output_image_path=os.path.join(output_dir,image_name)
    cv2.imwrite(output_image_path,resized_image)
        
resize_image(image_path,output_dir)   
```

```python
#observing the image and the label
import cv2
from PIL import Image
image = cv2.imread ("./archive/tfresized_data/images/train/IMG_0710.jpg")

with open("./archive/labels/train/IMG_0710.txt") as f:
    lines = f.read().strip().split('\n')
    for line in lines:
        values = line.split()
        class_id = int (values[0])
        center_x=float(values[1])
        center_y=float(values[2])
        width = float(values[3])
        height = float(values[4])

        #calculating cordinates
        image_height,image_width = image.shape[:2]
        x1=int((center_x-width/2)*image_width)#top_left
        y1=int((center_y-height/2)*image_height)
        x2=int((center_x+width/2)*image_width)
        y2=int((center_y+height/2)*image_height)#bottom_right

        cv2.rectangle(image,(x1,y1),(x2,y2),(255,0,0),1)#RGB,thikness

Image.fromarray(cv2.resize(image[:,:,::-1], (300,300)))
        
    
```
