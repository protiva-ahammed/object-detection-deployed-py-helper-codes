```python
import cv2
from PIL import Image
image =  cv2.imread("./archive/resized640yolo/images/train/IMG_0495.jpg")
with open ("./archive/resized640yolo/labels/train/IMG_0495.txt") as f:
    lines = f.read().strip().split("\n")
    for line in lines:
        
        #spliting lines by space to separate values
        values = line.split()
        #converting values in flot
        class_id = int(values[0])
        center_x = float(values[1])
        center_y = float(values[2])
        width = float(values[3])
        height = float(values[4])

        #calculating coordinates

        image_height,image_width = image.shape[:2]
        x1=int((center_x-width/2)*image_width)#top_left
        y1=int((center_y-height/2)*image_height)
        x2=int((center_x+width/2)*image_width)
        y2=int((center_y+height/2)*image_height)#bottom_right

        cv2.rectangle(image,(x1,y1),(x2,y2),(255,0,0),1)#RGB,thikness

Image.fromarray(cv2.resize(image[:,:,::-1], (640,640)))#color channel is reversed cz opencv reads BGR format and PIL reads RGB modifiying as array

    
```
