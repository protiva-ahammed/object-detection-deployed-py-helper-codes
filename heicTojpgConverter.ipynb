```python
#code to show case only one image reshaped and converted into .jpg
from PIL import Image
import pyheif
import pathlib

#open the HEIC file
heic_path = pathlib.Path('./archive/IMG_0433.HEIC')
heic_file = pyheif.read(heic_path)

#convertig image rgb
image = Image.frombytes(
    heic_file.mode,
    heic_file.size,
    heic_file.data,
    "raw",
    heic_file.mode,
    heic_file.stride,
)
#color space pixel format ==mode && stride specifies the number of bytes per row in the image data
#resize to nas format
image_resized = image.resize((640,640),Image.ANTIALIAS)
jpg_path = heic_path.with_suffix('.jpg')
image_resized.save(jpg_path)
```

```python

#code to resize and save all the images
from PIL import Image
import pyheif
# import pathlib
import os

input_dir = './archive/images/val1'
output_dir = './archive/images/val1/jpgImage'
os.makedirs(output_dir,exist_ok=True)

for filename in os.listdir(input_dir):
    if filename.endswith(".HEIC"):
        heic_path = os.path.join(input_dir,filename)
        heic_file = pyheif.read(heic_path)

        #rgb
        image = Image.frombytes(
            heic_file.mode,
            heic_file.size,
            heic_file.data,
            "raw",
            heic_file.mode,
            heic_file.stride,
        )#strides no of bytes between adjacent rows of pixel in a memory
        #resize the image
        image_resized = image.resize((640,640),Image.ANTIALIAS)

        #save as JPG
        jpg_filename = os.path.splitext(filename)[0] + '.jpg'
        jpg_path = os.path.join(output_dir,jpg_filename)
        print("Saving in path",jpg_filename)
        image_resized.save(jpg_path)

        
```
